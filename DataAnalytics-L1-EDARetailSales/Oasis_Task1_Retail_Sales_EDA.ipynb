```python
#4. Time Series Analysis

This section analyzes monthly and quarterly sales trends to 
understand how sales performance changes over time.
```

```python
##Monthly Sales
dataset["Date_of_Sale"]=pd.to_datetime(dataset['Date_of_Sale'])
dataset['Date_of_Sale'].head()
```

```python
monthly_sales = dataset.groupby(
    dataset['Date_of_Sale'].dt.to_period('M')
)['Sales_Amount'].sum()

monthly_sales
```

```python
monthly_sales.plot(
    kind='line',
    figsize=(12, 5),
    marker='o',
    title='Monthly Sales Trend'
)

plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid()
plt.show()
```

```python
# Monthly Observation

Sales remained relatively stable from January to August 2024. 
However, September recorded a significant decline in sales, with 
sales of approximately 1.96 million compared with approximately 6.05 million
in August.
```

```python
# Quarterly Sales Trend
```

```python
quarterly_sales = dataset.groupby(
    dataset['Date_of_Sale'].dt.to_period('Q')
)['Sales_Amount'].sum()

quarterly_sales

```

```python
quarterly_sales.plot(
    kind='line',
    figsize=(10, 5),
    marker='o',
    title='Quarterly Sales Trend'
)

plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.grid()
plt.show()
```

```python
# Quarterly Observation

recorded the highest quarterly sales at approximately 17.76 million,
followed by Q1 at approximately 17.66 million. Q3 recorded lower sales of
approximately 14.10 million, mainly influenced by the significant decline
in September.
```

```python
# 5. Customer Demographics

This section analyzes customer age groups and gender distribution to 
understand the characteristics of the customer base.
```

```python
##Age Group Analysis
bins = [0, 18, 25, 35, 45, 55, 100]

labels = ['Under 18', '18-25', '26-35', '36-45', '46-55', '55+']

dataset['Age_Group'] = pd.cut(
    dataset['Customer_Age'],
    bins=bins,
    labels=labels
)
dataset['Age_Group'].value_counts().sort_index()
```

```python
dataset['Customer_Gender'].value_counts()
```

```python
age_group_counts = dataset['Age_Group'].value_counts().sort_index()

age_group_counts.plot(
    kind='bar',
    figsize=(10, 5),
    title='Customer Distribution by Age Group'
)

plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
##Gender Analysis
gender_counts = dataset['Customer_Gender'].value_counts()

gender_counts.plot(
    kind='bar',
    figsize=(7, 5),
    title='Customer Distribution by Gender'
)

plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
### Observation

The age-group and gender charts provide an overview of the customer demographics.
This information can help the business understand its customer base and design
more targeted marketing strategies.
```

```python
## 6. Product Category Analysis

This section analyzes revenue generated by different product categories to 
identify high-performing and low-performing categories.
```

```python
#Revenue by Category
category_revenue = dataset.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)

category_revenue
```

```python
#Bar Chart
category_revenue.plot(
    kind='bar',
    figsize=(10, 5),
    title='Revenue by Product Category'
)

plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
### Observation

Movies generated the highest revenue at approximately 2.13 million, followed by
Toys and Books. Automotive and Software generated comparatively lower revenue among
the product categories. The business can focus on high-performing categories while 
reviewing pricing, promotions, and customer demand for lower-performing categories.
```

```python
###Correlation Analysis

This section examines the correlation between numerical variables in the retail 
sales dataset.
```

```python
##Correlation Matrix
correlation = dataset.corr(numeric_only=True)

correlation
```

```python
#heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
### Observation
The correlation heatmap shows the relationships between numerical variables.
Stronger positive or negative correlation values indicate a stronger 
relationship between two variables.
```

```python
### Additional Visualization

This visualization explores the relationship between discount and sales amount
to identify patterns and potential outliers.
```

```python
##Scatter Plot
plt.figure(figsize=(10, 6))

plt.scatter(
    dataset['Discount'],
    dataset['Sales_Amount'],
    alpha=0.5
)

plt.title('Discount vs Sales Amount')
plt.xlabel('Discount')
plt.ylabel('Sales Amount')

plt.show()
```

```python
### Observation

The scatter plot shows the relationship between discount and sales amount. 
The distribution of points helps identify whether different discount levels are
associated with changes in sales amount and also highlights potential outliers.
```

```python
### Observation

The scatter plot shows the relationship between discount and sales amount. 
It helps identify whether higher discounts are associated with higher or lower sales
amounts and reveals any unusual patterns or outliers.
```
